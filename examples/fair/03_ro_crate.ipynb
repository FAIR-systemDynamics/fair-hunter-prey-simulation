# Package an actual execution

Select the implementation and case explicitly. The crate contains the selected model, original inputs, preserved runner/readers, native reference, actual PySD output, exact environment, hashes, citation and per-file licensing. This example creates local files; it does not upload, mint a DOI, or activate a webhook. Model-derived material is CC BY-NC-SA 4.0; repository software is MIT.

```python
from pathlib import Path
from tempfile import mkdtemp
import json
import hashlib
from fair_hunter_prey import run_example, create_crate
implementation = "stella"
case = "case2"
output = Path(mkdtemp(prefix="hunter-prey-crate-"))
run = run_example(implementation, case, output / "run")
crate = create_crate(output / "run", output / "crate")
print(crate)
```

## Inspect the record

The execution engine is PySD. Stella Case 2’s native reference contains final values, while `execution/simulation.csv` contains the trajectory newly generated by PySD. A portable `reproduce.py` writes a separate `reproduced/` directory.

```python
record = json.loads(run["provenance"].read_text())
metadata = json.loads((crate / "ro-crate-metadata.json").read_text())
assert record["engine"] == "PySD"
for entity in metadata["@graph"]:
    if "sha256" in entity:
        assert hashlib.sha256((crate / entity["@id"]).read_bytes()).hexdigest() == entity["sha256"]
print("Verified payload hashes;", len(metadata["@graph"]), "metadata entities")
print((crate / "README-crate.txt").read_text())
```
